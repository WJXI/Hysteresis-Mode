```python
#       Fig 1 C&D

import numpy as np
import scipy.sparse as sp
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def plot_flux_comparison_by_time_optimized(
    skeleton_ids,      
    sim_data,          
    mn9_id,
    time_range_A=(1500, 2000),    # feeding state A
    time_range_B=(6000, 6500),    # escape state B
    remove=[],          # [SUGAR, ESC]
    

    label_A="Feeding State (Maintenance)",
    label_B="Suppression State (Escape)",
    main_title="Functional Reconfiguration under Conflict",
    save_filename="Figure2_Functional_Reconfiguration"
):
    print(f"=== Visualizing Flux: {label_A} vs {label_B} ===")
    
    # --- 1.  ---
    full_id_map = sim_data['id_map']
    flux_list = sim_data['flux_list']
    time_list = np.array(sim_data['time_list'])
    
    try:
        skel_indices_in_full = [full_id_map[uid] for uid in skeleton_ids]
    except KeyError as e:
        print(f"Error: Skeleton ID {e} not found.")
        return

    ids_to_remove = set()
    if remove:
        for item in remove:
            if np.ndim(item) > 0: ids_to_remove.update(item)
            else: ids_to_remove.add(item)
            
    skel_arr = np.array(skeleton_ids)
    mask_keep = ~np.isin(skel_arr, list(ids_to_remove))
    
    plot_ids = skel_arr[mask_keep]
    plot_indices = np.array(skel_indices_in_full)[mask_keep]
    N_core = len(plot_ids)
    
    print(f"  > Core Neurons to Plot: {N_core}")
    if N_core == 0: return

    # --- 2. ---
    def get_avg_slice_by_time(t_min, t_max):
        mask = (time_list >= t_min) & (time_list <= t_max)
        target_indices = np.where(mask)[0]
        
        if len(target_indices) == 0:
            print(f"    ! Warning: No data for {t_min}-{t_max} ms.")
            return np.zeros((N_core, N_core))
            
        accumulated_F = np.zeros((N_core, N_core))
        for i in target_indices:
            F_full = flux_list[i]
            if sp.issparse(F_full):
                F_sub = F_full[plot_indices, :][:, plot_indices]
                accumulated_F += F_sub.toarray()
            else:
                F_sub = F_full[np.ix_(plot_indices, plot_indices)]
                accumulated_F += F_sub
        return accumulated_F / len(target_indices)

    # --- 3.  ---
    print("  > Extracting matrices...")
    F_A = get_avg_slice_by_time(*time_range_A)
    F_B = get_avg_slice_by_time(*time_range_B)
    
    # --- 4. ---
    print("  > Clustering based on State A...")
    F_ref = np.abs(F_A)
    # 
    if np.max(F_ref) == 0: F_ref = np.abs(F_B)
        
    Sim = F_ref + F_ref.T
    try:
        from scipy.spatial.distance import squareform
        Dist = np.max(Sim) - Sim
        np.fill_diagonal(Dist, 0)
        linkage = sch.linkage(squareform(Dist, checks=False), method='ward')
        sort_idx = sch.leaves_list(linkage)
    except:
        sort_idx = np.arange(N_core)
        
    # --- 5. ---
    def symlog(F):
        return np.sign(F) * np.log1p(np.abs(F))
    
    D1 = symlog(F_A)
    D2 = symlog(F_B)
    
    #
    all_vals = np.concatenate([np.abs(D1).flatten(), np.abs(D2).flatten()])
    all_vals = all_vals[all_vals > 1e-6]
    vmax = np.percentile(all_vals, 99) if len(all_vals) > 0 else 1.0
    
    #
    try:
        mn9_loc = np.where(plot_ids == mn9_id)[0][0]
        mn9_plot = np.where(sort_idx == mn9_loc)[0][0]
    except: mn9_plot = None

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))
    
    def draw(ax, data, title_str):
        d_sort = data[np.ix_(sort_idx, sort_idx)]
        im = ax.imshow(d_sort, cmap='RdBu_r', vmin=-vmax, vmax=vmax, interpolation='nearest')
        
        #
        ax.set_title(title_str, fontsize=16, fontweight='bold', pad=15)
        ax.set_xticks([])
        ax.set_yticks([])
       
        
        if mn9_plot is not None:
            c = 'limegreen'
            ax.add_patch(patches.Circle((mn9_plot, mn9_plot), N_core*0.02, fc='none', ec=c, lw=2))
            ax.axvline(mn9_plot, c=c, ls=':', alpha=0.5)
            ax.axhline(mn9_plot, c=c, ls=':', alpha=0.5)
            # 
            if ax == ax1:
                ax.text(mn9_plot+N_core*0.03, mn9_plot-N_core*0.03, "MN9", 
                        color=c, fontweight='bold', fontsize=12,
                        bbox=dict(facecolor='white', alpha=0.7, ec='none', pad=1))
        return im

    # 
    im1 = draw(ax1, D1, label_A)
    im2 = draw(ax2, D2, label_B)
    
    # Colorbar
    cbar = fig.colorbar(im1, ax=[ax1, ax2], fraction=0.025, pad=0.04)
    cbar.set_label('Log Signal Flux (Excitation / Inhibition)', rotation=270, labelpad=20)
    
    #
    
    # --- 6. 保存 ---
    print(f"  > Saving to {save_filename}.pdf/.png ...")
    plt.savefig(f"{save_filename}.pdf", bbox_inches='tight')
    plt.savefig(f"{save_filename}.png", dpi=300, bbox_inches='tight')
    
    plt.show()

# ==========================================
# 调用示例
# ==========================================
if __name__ == "__main__":
    try:
        plot_flux_comparison_by_time_optimized(
            skeleton_ids=skel_ids, 
            sim_data=sim_data, 
            mn9_id=MN9,
            time_range_A=(1500, 2000), #  High Sugar
            time_range_B=(6000, 6500), #  High Threat
            remove=[SUGAR, ESC],
            label_A="(C) Feeding State",
            label_B="(D) Escape State",
            main_title="Structural Basis of Hysteresis",
            save_filename="Fig1(CD)"
        )
    except NameError:
        pass
```

```python
# find feeding core
import numpy as np
import scipy.sparse as sp
from scipy.sparse.csgraph import connected_components

def analyze_attractor_ensemble_v2(
    sim_data, 
    mn9_id,
    flux_threshold_percent=5.0, 
    time_window=(1000, 2000)
):
    print(f"=== Attractor Ensemble Analysis (Returns Core IDs) ===")
    
    # --- 1. Data ---
    flux_list = sim_data['flux_list']
    time_list = np.array(sim_data['time_list'])
    neuron_ids = np.array(sim_data['neuron_ids'])
    id_map = sim_data['id_map']
    W_csr = sim_data['W_csr']
    
    try: mn9_idx = id_map[mn9_id]
    except: return None, None

    #  Flux
    t_start, t_end = time_window
    mask = (time_list >= t_start) & (time_list <= t_end)
    indices = np.where(mask)[0]
    
    if len(indices) == 0:
        print("Error: Time window empty.")
        return None, None

    F_accum = None
    for i in indices:
        F = flux_list[i]
        if sp.issparse(F): F = F.toarray()
        if F_accum is None: F_accum = np.abs(F)
        else: F_accum += np.abs(F)
    F_avg_abs = F_accum / len(indices)
    

    W_sign = np.sign(W_csr.toarray())
    F_avg_signed = F_avg_abs * W_sign 

    # --- 2. Find Attractor ---
    limit = np.max(F_avg_abs) * (flux_threshold_percent / 100.0)
    Adj = (F_avg_abs > limit).astype(int)
    n_comp, labels = connected_components(sp.csr_matrix(Adj), connection='strong')
    
    unique_labels, counts = np.unique(labels, return_counts=True)
    sorted_idx = np.argsort(counts)[::-1]
    
    valid_attractors = [] 
    
    print(f"  > Scanning candidates (Threshold > {limit:.2f})...")
    print(f"  {'ID':<4} {'Size':<6} {'Exc%':<8} {'Status'}")
    print("-" * 40)
    
    for idx in sorted_idx:
        lbl = unique_labels[idx]
        size = counts[idx]
        if size < 2: continue 
        
        member_indices = np.where(labels == lbl)[0]
        
        # E/I ratio
        W_sub = W_csr[np.ix_(member_indices, member_indices)]
        F_sub = F_avg_abs[np.ix_(member_indices, member_indices)]
        valid_edges = F_sub > limit
        if np.sum(valid_edges) > 0:
            edge_signs = np.sign(W_sub.toarray()[valid_edges])
            exc_ratio = np.sum(edge_signs > 0) / len(edge_signs) * 100
        else: exc_ratio = 0
        
        if exc_ratio > 60.0: 
            status = "✅ Valid"
            valid_attractors.append({
                'label': lbl,
                'size': size,
                'indices': member_indices,
                'exc_ratio': exc_ratio,
                'drive': 0.0 
            })
        else:
            status = "❌ Unstable"
            
        print(f"  {lbl:<4} {size:<6} {exc_ratio:>5.1f}%   {status}")
        
    print(f"\n  > Found {len(valid_attractors)} Valid Attractor Modules.")
    
    # --- 3. Find Best Core ---
    print(f"\n[Drive Analysis by Module]")
    
    F_pos = F_avg_signed.copy(); F_pos[F_pos<0] = 0
    
    best_core_ids = []
    max_drive = -1.0
    best_module_idx = -1
    
    for i, mod in enumerate(valid_attractors):
        indices = mod['indices']
        

        drive = np.sum(F_pos[mn9_idx, indices])
        mod['drive'] = drive 
        
        print(f"  Module {mod['label']} (Size {mod['size']}): Drive = {drive:.1f}")
        
        # Find powerful Core
        if drive > max_drive:
            max_drive = drive
            best_module_idx = i

            best_core_ids = neuron_ids[indices]

    if best_module_idx != -1:
        best_mod = valid_attractors[best_module_idx]
        print(f"\n  >>> PRIMARY ENGINE IDENTIFIED: Module {best_mod['label']} (Size {best_mod['size']})")
        print(f"      Drive Strength: {max_drive:.1f}")
        print(f"      Core IDs extracted: {len(best_core_ids)} neurons")
    else:
        print("\n  ! Warning: No module drives MN9 significantly.")

        if len(valid_attractors) > 0:
             best_core_ids = neuron_ids[valid_attractors[0]['indices']]
             print("      Fallback: Returning largest valid module.")

    return valid_attractors, list(best_core_ids)
    

# ==========================================

# ==========================================
if __name__ == "__main__":

    att_info, core_ids_18 = analyze_attractor_ensemble_v2(sim_data, MN9)
```

```python
# Fig3(A)

import numpy as np
import scipy.sparse as sp
import matplotlib.pyplot as plt
from scipy.sparse.csgraph import connected_components

def plot_attractor_drive_vs_rate(
    sim_data, 
    mn9_id,
    flux_threshold_percent=5.0,
    feeding_window=(1500, 2500)
):
    print(f"=== Dynamics: Attractor Drive vs MN9 Rate ===")
    

    flux_list = sim_data['flux_list']
    time_list = np.array(sim_data['time_list'])
    W_csr = sim_data['W_csr']
    id_map = sim_data['id_map']
    neuron_ids = np.array(sim_data['neuron_ids'])
    
    try: mn9_idx = id_map[mn9_id]
    except: 
        print("Error: MN9 not found in ID map.")
        return


    print("  > Identifying the Attractor (Feeding Phase)...")
    

    mask_feed = (time_list >= feeding_window[0]) & (time_list <= feeding_window[1])
    indices_feed = np.where(mask_feed)[0]
    
    if len(indices_feed) == 0:
        print(f"Error: No data in feeding window {feeding_window}. Check sim_data['time_list'].")
        return

    F_accum = None
    for i in indices_feed:
        F = flux_list[i]
        if sp.issparse(F): F = F.toarray()
        if F_accum is None: F_accum = np.abs(F)
        else: F_accum += np.abs(F)
    F_avg_abs = F_accum / len(indices_feed)
    

    limit = np.max(F_avg_abs) * (flux_threshold_percent / 100.0)
    Adj = (F_avg_abs > limit).astype(int)
    n_comp, labels = connected_components(sp.csr_matrix(Adj), connection='strong')
    
    unique_labels, counts = np.unique(labels, return_counts=True)
    sorted_scc_idx = np.argsort(counts)[::-1]
    

    best_label = unique_labels[sorted_scc_idx[0]]
    attractor_indices = np.where(labels == best_label)[0]
    
    print(f"  > Locked on Attractor: Size {len(attractor_indices)}")
    

    print("  > Extracting time series data...")
    
    trace_drive = [] 
    trace_rate = []  
    

    mn9_out_weights = np.sum(np.abs(W_csr[:, mn9_idx].toarray()))
    
    for i, t in enumerate(time_list):
        F_t = flux_list[i]
        if sp.issparse(F_t): F_t = F_t.toarray()
        
        # A. Attractor -> MN9 Flux
 
        input_flux = np.sum(F_t[mn9_idx, attractor_indices])
        physical_flux = input_flux * 0.275  # <--- 补上这个物理常数
        trace_drive.append(max(0, physical_flux))
        
        # B. MN9 rate
        # nu = Total_Output_Flux / Total_Output_Weight
        total_out_flux = np.sum(np.abs(F_t[:, mn9_idx]))
        
        rate = total_out_flux / mn9_out_weights if mn9_out_weights > 0 else 0
        trace_rate.append(rate)
        
    # --- 4. plot---
    fig, ax1 = plt.subplots(figsize=(12, 6))
    
    color_drive = 'firebrick'
    color_rate = 'darkgreen'
    
    # Left ax：Flux Drive
    ax1.set_xlabel('Time (ms)', fontsize=12)
    ax1.set_ylabel('MN9 Firing Rate (Hz)', color=color_rate, fontsize=12, fontweight='bold')
    l1, = ax1.plot(time_list, trace_drive, color=color_drive, linewidth=2.5, alpha=0.8, label='Core Flux to MN9(mV/s)')
    ax1.tick_params(axis='y', labelcolor=color_drive)
    ax1.grid(True, linestyle='--', alpha=0.4)
    
    # Right ax：MN9 Rate
    ax2 = ax1.twinx() 
    ax2.set_ylabel('Core Flux to MN9(mV/s)', color=color_drive, fontsize=12, fontweight='bold')
    l2, = ax2.plot(time_list, trace_rate, color=color_rate, linewidth=2.5, linestyle=':', label='MN9 Firing Rate')
    ax2.tick_params(axis='y', labelcolor=color_rate)
    

    try:
        t1, t2, t3, t4 = sim_data['time_markers']
        ax1.axvspan(t2, t3, color='gray', alpha=0.1, label='Threat Ramp')
    except (KeyError, ValueError): pass
    
    plt.title(f"(A)Coupling Dynamics", fontsize=14, fontweight='bold', pad=15)
    

    lines = [l1, l2]
    labels = [l.get_label() for l in lines]
    ax1.legend(lines, labels, loc='upper right')
    
    plt.tight_layout()
    

    filename = "Fig3(A)"
    try:
        plt.savefig(f"{filename}.pdf", bbox_inches='tight')
        plt.savefig(f"{filename}.png", dpi=300, bbox_inches='tight')
        print(f"\n✅ Figures saved as '{filename}.pdf' and '{filename}.png'")
    except Exception as e:
        print(f"\n❌ Error saving figure: {e}")

    plt.show()

# ==========================================
# 运行
# ==========================================
if __name__ == "__main__":
    try:
        plot_attractor_drive_vs_rate(
    sim_data, 
    MN9,
    flux_threshold_percent=5.0,
    feeding_window=(1000, 2000)
)
    except NameError:
        print("Required variables (sim_data, MN9) not found.")
```

```python
import brian2 as b2
import numpy as np
import scipy.sparse as sp
import matplotlib.pyplot as plt

# =============================================================================
# 1. single simulation (Magic Mode)
# =============================================================================
def run_single_decay_trial(
    run_id,
    neuron_ids, W_csr, sugar_ids, mn9_idx, core_indices,
    dt_sim, input_gain,
    ramp_time, hold_time, decay_time, sugar_freq_max,
    plot_dt=10.0, window_ms=50.0
):
    print(f"  > Trial {run_id+1} running...", end="\r")
    
    b2.start_scope()
    b2.defaultclock.dt = dt_sim * b2.ms
    
    N = len(neuron_ids)
    

    t1 = ramp_time
    t2 = t1 + hold_time # Cutoff Point
    total_time = t2 + decay_time
    
    time_steps = np.arange(0, total_time, dt_sim)
    sugar_vals = np.zeros_like(time_steps)
    
    # Sugar Profile
    mask_ramp = time_steps < t1
    sugar_vals[mask_ramp] = sugar_freq_max * (time_steps[mask_ramp] / t1)
    mask_hold = (time_steps >= t1) & (time_steps < t2)
    sugar_vals[mask_hold] = sugar_freq_max

    
    sugar_timed = b2.TimedArray(sugar_vals*b2.Hz, dt=dt_sim*b2.ms)
    

    tau_m = 20*b2.ms
    w_val = 0.275*b2.mV * input_gain
    eqs = '''
    dv/dt=(-v+g)/(20*ms):volt(unless refractory)
    dg/dt=-g/(5*ms):volt
    ref_P:second
    '''
    neurons = b2.NeuronGroup(N, eqs, threshold='v>7*mV', reset='v=0*mV', refractory='ref_P', method='exact')
    neurons.v=0; neurons.g=0; neurons.ref_P=2.2*b2.ms
    

    P_s = None; S_s = None
    if len(sugar_ids) > 0:
        neurons.ref_P[sugar_ids] = 0*b2.ms
        P_s = b2.PoissonGroup(len(sugar_ids), rates='sugar_timed(t)')
        S_s = b2.Synapses(P_s, neurons, on_pre='g+=w_val')
        S_s.connect(i=np.arange(len(sugar_ids)), j=sugar_ids)
    

    coo = W_csr.tocoo()
    S_rec = b2.Synapses(neurons, neurons, 'w:1', on_pre='g+=w*0.275*mV', delay=1.8*b2.ms)
    S_rec.connect(i=coo.col, j=coo.row)
    S_rec.w = coo.data
    
    spikes = b2.SpikeMonitor(neurons)
    
 
    b2.run(total_time * b2.ms, namespace=locals())
    

    all_t = spikes.t/b2.ms
    all_i = spikes.i
    plot_times = np.arange(0, total_time, plot_dt)
    trace_mn9_rate = []
    trace_core_flux = []
    
    weights_core = np.zeros(len(core_indices))
    for k, src in enumerate(core_indices): 
        weights_core[k] = W_csr[mn9_idx, src] * 0.275
        
    for t in plot_times:
        t_start = max(0, t - window_ms)
        mask = (all_t >= t_start) & (all_t < t)
        dur = window_ms/1000.0
        
        # MN9 Rate
        trace_mn9_rate.append(np.sum((all_i==mn9_idx)&mask)/dur)
        
        # Core Flux
        flux_t = 0
        if np.sum(mask) > 0:
            c = np.bincount(all_i[mask], minlength=N)
            c_core = c[core_indices]
            flux_t = np.sum(weights_core * (c_core/dur))
        trace_core_flux.append(flux_t)
        
    trace_mn9_rate = np.array(trace_mn9_rate)
    
    # calculate lifetime
    idx_cut = int(t2/plot_dt)

    base_window = int(200 / plot_dt) 
    base = np.mean(trace_mn9_rate[idx_cut-base_window:idx_cut])
    
    decay_part = trace_mn9_rate[idx_cut:]
    
    if base < 1.0: 
        lifetime = 0
    else:
        stop_indices = np.where(decay_part < base*0.1)[0]
        if len(stop_indices) > 0:
            lifetime = plot_times[idx_cut:][stop_indices[0]] - t2
        else:
            lifetime = decay_time
    
    return plot_times, trace_mn9_rate, np.array(trace_core_flux), lifetime, t2

# =============================================================================

# =============================================================================
def measure_attractor_lifetime_stats(
    neuron_ids, W_matrix, sugar_ids, mn9_id, core_ids,
    dt_sim=0.1, input_gain=200.0,
    ramp_time=500.0,    # warm up 500ms
    hold_time=1000.0,   # hold 1000ms
    decay_time=1000.0,  # decay 1000ms
    sugar_freq_max=100.0,
    n_trials=10
):
    print(f"=== Measuring Quasi-Particle Lifetime (N={n_trials}) ===")
    print(f"  > Protocol: Ramp({ramp_time}ms) -> Hold({hold_time}ms) -> Cutoff -> Decay({decay_time}ms)")
    
    id_map = {uid: i for i, uid in enumerate(neuron_ids)}
    try:
        mn9_idx = id_map[mn9_id]
        core_indices = np.array([id_map[uid] for uid in core_ids])
        s_indices = [id_map[uid] for uid in sugar_ids if uid in id_map]
    except KeyError:
        print("Error: IDs mismatch.")
        return

    W_csr = W_matrix.tocsr() if sp.issparse(W_matrix) else sp.csr_matrix(W_matrix)
    
    lifetimes = []
    mn9_matrix = []
    flux_matrix = []
    common_times = None
    cutoff_time = 0
    
    for i in range(n_trials):
        t, rate, flux, life, t_cut = run_single_decay_trial(
            i, neuron_ids, W_csr, s_indices, mn9_idx, core_indices,
            dt_sim, input_gain, ramp_time, hold_time, decay_time, sugar_freq_max
        )
        lifetimes.append(life)
        mn9_matrix.append(rate)
        flux_matrix.append(flux)
        if common_times is None:
            common_times = t
            cutoff_time = t_cut
            
    print(f"\n=== Statistics (N={n_trials}) ===")
    mean_life = np.mean(lifetimes)
    std_life = np.std(lifetimes)
    print(f"  > Lifetime: {mean_life:.0f} ± {std_life:.0f} ms")
    
    mn9_mean = np.mean(np.vstack(mn9_matrix), axis=0)
    mn9_std = np.std(np.vstack(mn9_matrix), axis=0)
    flux_mean = np.mean(np.vstack(flux_matrix), axis=0)
    
    # --- plot ---
    fig, ax1 = plt.subplots(figsize=(12, 6))
    

    ax1.set_xlim(0, common_times[-1])

    
    # 1. Ramp Up (Grey)
    ax1.axvspan(0, ramp_time, color='gray', alpha=0.1, label='Ramp Up (Charging)')
    # 2. Hold (Light Green / White) 
    ax1.axvspan(ramp_time, cutoff_time, color='green', alpha=0.05, label='Steady State')
    # 3. Lifetime (Gold) 
    ax1.axvspan(cutoff_time, cutoff_time + mean_life, color='gold', alpha=0.15, label='Mean Lifetime')
    
    # Cutoff line
    ax1.axvline(cutoff_time, color='k', linestyle='--', linewidth=2, label='Sugar Cutoff')
    
    # MN9 Rate (Green)
    c1 = 'tab:green'
    ax1.set_xlabel('Time (ms)', fontsize=12)
    ax1.set_ylabel('MN9 Firing Rate (Hz)', color=c1, fontsize=12, fontweight='bold')
    
    mn9_lower = np.clip(mn9_mean - mn9_std, 0, None)
    
    ax1.plot(common_times, mn9_mean, color=c1, lw=2.5, label='MN9 Rate (Mean)')
    ax1.fill_between(common_times, mn9_lower, mn9_mean + mn9_std, 
                     color=c1, alpha=0.2, label='MN9 Rate ±1 SD')
    
    ax1.tick_params(axis='y', labelcolor=c1)
    ax1.set_ylim(bottom=0)
    
    # Core Flux (Red)
    ax2 = ax1.twinx()
    c2 = 'tab:red'
    ax2.set_ylabel('Core Flux Output (mV/s)', color=c2, fontsize=12, fontweight='bold')
    
    ax2.plot(common_times, flux_mean, color=c2, lw=2, linestyle='--', label='Core Flux (Mean)')
    
    ax2.tick_params(axis='y', labelcolor=c2)
    ax2.set_ylim(bottom=0)
    

    text_x = cutoff_time + mean_life / 2
    text_y = np.max(mn9_mean) * 0.85
    
    ax1.text(text_x, text_y, 
             f'Lifetime\n{mean_life:.0f} ± {std_life:.0f} ms', 
             ha='center', va='center', color='black', fontsize=11, fontweight='bold',
             bbox=dict(facecolor='white', alpha=0.9, edgecolor='gray', pad=4))

    #plt.title(f"Full Dynamics: Charging, Holding, and Decay (N={n_trials})", fontsize=14, pad=15)
    plt.title(f"(B)Life time of the metastable feeding mode", fontsize=14, fontweight='bold', pad=15)
    # 图例
    lines1, labels1 = ax1.get_legend_handles_labels()
    lines2, labels2 = ax2.get_legend_handles_labels()
    ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', frameon=True, fontsize=10, ncol=2)
    
    plt.tight_layout()
    plt.savefig('Fig3(B).pdf')
    plt.savefig('Fig3(B).png', dpi=300)
    plt.show()

# ==========================================
# run
# ==========================================
if __name__ == "__main__":
    try:
        measure_attractor_lifetime_stats(
            skel_ids, skel_W, SUGAR, MN9, core_ids_18, 
            n_trials=10,
            ramp_time=500,     # 0-500ms 预热
            hold_time=1000,    # 500-1500ms 稳定
            decay_time=1000    # 1500-2500ms 衰减
        )
    except: pass
```

```python
# Find escape core
import numpy as np
import scipy.sparse as sp
import matplotlib.pyplot as plt
from scipy.sparse.csgraph import connected_components

def quantify_escape_cascade_custom(
    sim_data, 
    feeding_core_ids,  # feeding core (Size 18)
    suppress_window=(6000, 7000),  # escape phase
    top_k_killers=20,              # num of killer
    top_k_commanders=20            # num of commander
):
    print(f"=== Quantifying Escape Cascade (with Topological Refinement) ===")
    print(f"  > Target (Victim): Feeding Core (Size {len(feeding_core_ids)})")
    

    flux_list = sim_data['flux_list']
    time_list = np.array(sim_data['time_list'])
    neuron_ids = np.array(sim_data['neuron_ids'])
    id_map = sim_data['id_map']
    W_csr = sim_data['W_csr']
    

    try:
        feeding_core_indices = np.array([id_map[uid] for uid in feeding_core_ids])
    except KeyError as e:
        print(f"Error: ID {e} not found.")
        return None, None

    # Flux 
    t_start, t_end = suppress_window
    mask = (time_list >= t_start) & (time_list <= t_end)
    indices = np.where(mask)[0]
    
    if len(indices) == 0:
        print(f"Error: No data in window {suppress_window}")
        return None, None

    F_accum = None
    for i in indices:
        F = flux_list[i]
        if sp.issparse(F): F = F.toarray()
        if F_accum is None: F_accum = np.abs(F)
        else: F_accum += np.abs(F)
    F_avg_abs = F_accum / len(indices)
    
    # sign
    W_sign = np.sign(W_csr.toarray())
    F_signed = F_avg_abs * W_sign 

    # =========================================================================
    # STEP 1:  (Find Killers)
    # =========================================================================
    print("\n[Step 1] Identifying Killers (Direct Inhibitors)...")
    
    # Flux to Feeding Core 
    inputs_to_core = np.sum(F_signed[feeding_core_indices, :], axis=0)
    

    sorted_idx = np.argsort(inputs_to_core) 
    
    killer_indices = []
    count = 0
    for idx in sorted_idx:
        val = inputs_to_core[idx]
        if val < -100: 
            if idx not in feeding_core_indices:
                killer_indices.append(idx)
                count += 1
        else:
            break 
        if count >= top_k_killers: break
            
    print(f"  > Killers Identified: {len(killer_indices)}")
    
    # =========================================================================
    # STEP 2:  (Find Commander Candidates)
    # =========================================================================
    print("\n[Step 2] Identifying Commander Candidates...")
    
    commander_candidates = []
    if len(killer_indices) > 0:
        inputs_to_killers = np.sum(F_signed[killer_indices, :], axis=0)
        sorted_cmd = np.argsort(inputs_to_killers)[::-1] 
        
        threat_set = set([id_map[u] for u in sim_data['threat_ids'] if u in id_map])
        
        count = 0
        for idx in sorted_cmd:
            val = inputs_to_killers[idx]
            if val > 100: 
                if (idx not in threat_set) and (idx not in killer_indices) and (idx not in feeding_core_indices):
                    commander_candidates.append(idx)
                    count += 1
            else: break
            if count >= top_k_commanders: break
    
    print(f"  > Candidates found: {len(commander_candidates)}")

    # =========================================================================
    # STEP 3:  (Refining the Core)
    # =========================================================================
    print("\n[Step 3] Topological Refinement (Looking for Recurrent Loop)...")
    
    if len(commander_candidates) > 1:

        W_cmd = W_csr[np.ix_(commander_candidates, commander_candidates)]
        

        W_cmd_pos = W_cmd.copy()
        W_cmd_pos.data = np.where(W_cmd_pos.data > 0, W_cmd_pos.data, 0)
        W_cmd_pos.eliminate_zeros()
        

        n_comp, labels = connected_components(W_cmd_pos, connection='strong')
        unique, counts = np.unique(labels, return_counts=True)
        

        max_size = np.max(counts)
        best_label = unique[np.argmax(counts)]
        
        if max_size > 1:
            core_mask = (labels == best_label)
            escape_core_indices = np.array(commander_candidates)[core_mask]
            print(f"  > ✅ Recurrent Escape Core Detected: Size {max_size}")
            print(f"  > (Filtered out {len(commander_candidates) - max_size} feedforward drivers)")
        else:
            print("  > ❌ No Recurrent Core found. Using all candidates.")
            escape_core_indices = commander_candidates
    else:
        escape_core_indices = commander_candidates

    # =========================================================================
    # STEP 4: Plot
    # =========================================================================
    print("\n[Step 4] Quantifying Cascade Strength...")
    
    # 1.  (Self-Flux of Escape Core)
    flux_self = np.sum(F_signed[np.ix_(escape_core_indices, escape_core_indices)])
    
    # 2.  (Escape Core -> Killers)
    flux_cmd = np.sum(F_signed[np.ix_(killer_indices, escape_core_indices)])
    
    # 3.  (Killers -> Feeding Core)
    flux_kill = np.sum(F_signed[np.ix_(feeding_core_indices, killer_indices)])
    
    print(f"  > Flux Cascade: Self({flux_self:.0f}) -> Cmd({flux_cmd:.0f}) -> Kill({flux_kill:.0f})")


    fig, ax = plt.subplots(figsize=(8, 6))
    
    stages = ['Escape Core\n(Self-Maintenance)', 'Command Drive\n(Core -> Killers)', 'Suppression\n(Killers -> Feeding)']
    values = [flux_self, flux_cmd, flux_kill]
    colors = ['firebrick', 'darkorange', 'royalblue']
    
    bars = ax.bar(stages, values, color=colors, alpha=0.85, edgecolor='black', linewidth=1.5)
    
    ax.axhline(0, color='black', linewidth=1)
    ax.set_ylabel("Net Signal Flux (mV/s)", fontsize=12, fontweight='bold')
    ax.set_title(f"Escape Cascade Dynamics\n(Core Size: {len(escape_core_indices)} | Killer Size: {len(killer_indices)})", fontsize=14, pad=15)
    

    y_min = min(values) * 1.2
    y_max = max(max(values) * 1.2, abs(min(values))*0.2)
    ax.set_ylim(y_min, y_max)
    
    for bar in bars:
        h = bar.get_height()
        label_y = h + (y_max - y_min)*0.05 if h >= 0 else h - (y_max - y_min)*0.05
        ax.text(bar.get_x() + bar.get_width()/2, label_y, f"{h:.0f}", 
                ha='center', va='center', fontweight='bold', fontsize=11)
        
    plt.tight_layout()
    plt.savefig('Figure5_Cascade_Refined.pdf')
    plt.show()
    
    return neuron_ids[escape_core_indices], neuron_ids[killer_indices]

# ==========================================

# ==========================================
if __name__ == "__main__":
    try:

        esc_ids, kill_ids = quantify_escape_cascade_custom(
            sim_data, 
            feeding_core_ids=core_ids_18,
            top_k_killers=10,
            top_k_commanders=20 
        )
    except NameError:
        pass
```

```python
# Fig2
import numpy as np
import scipy.sparse as sp
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def compare_active_cores_final(
    sim_data, 
    feeding_core_ids,  
    escape_core_ids,   
    mn9_id,
    feed_window=(1500, 2000),  
    suppress_window=(6000, 6500), 
    save_name="Figure_Core_Comparison"
):
    print(f"=== Visualizing Active Cores: Feeding vs Escape ===")
    

    W_full = sim_data['W_csr']
    flux_list = sim_data['flux_list']
    time_list = np.array(sim_data['time_list'])
    id_map = sim_data['id_map']
    

    def process_core_data(core_ids, time_window):

        try:
            indices = [id_map[uid] for uid in core_ids]
        except KeyError:
            print("Error: IDs not found.")
            return None
        
        N = len(indices)
        

        W_sub = W_full[np.ix_(indices, indices)].toarray()
        

        mask = (time_list >= time_window[0]) & (time_list <= time_window[1])
        t_indices = np.where(mask)[0]
        
        if len(t_indices) == 0:
            F_sub = np.zeros_like(W_sub)
        else:
            F_accum = np.zeros_like(W_sub)
            for i in t_indices:
                F_frame = flux_list[i]
                if sp.issparse(F_frame):
   
                    F_sub_frame = F_frame[np.ix_(indices, indices)].toarray()
                else:
                    F_sub_frame = F_frame[np.ix_(indices, indices)]
                F_accum += F_sub_frame
            F_sub = F_accum / len(t_indices)
            

        F_abs = np.abs(F_sub)

        Sort_Base = F_abs + np.abs(W_sub) * 0.01 
        
        Sim = Sort_Base + Sort_Base.T
        np.fill_diagonal(Sim, np.max(Sim))
        
        try:
            from scipy.spatial.distance import squareform
            Dist = np.max(Sim) - Sim
            np.fill_diagonal(Dist, 0)
            linkage = sch.linkage(squareform(Dist, checks=False), method='ward')
            sort_idx = sch.leaves_list(linkage)
        except:
            sort_idx = np.arange(N)
            

        W_sorted = W_sub[np.ix_(sort_idx, sort_idx)]
        F_sorted = F_sub[np.ix_(sort_idx, sort_idx)]
        

        mn9_pos = None
        if mn9_id in core_ids:
            raw_pos = list(core_ids).index(mn9_id)
            mn9_pos = np.where(sort_idx == raw_pos)[0][0]
            
        return W_sorted, F_sorted, mn9_pos, N, sort_idx


    print(f"  > Processing Feeding Core (Window: {feed_window} ms)...")
    data_feed = process_core_data(feeding_core_ids, feed_window)
    
    print(f"  > Processing Escape Core (Window: {suppress_window} ms)...")
    data_esc = process_core_data(escape_core_ids, suppress_window)
    
    if data_feed is None or data_esc is None: return

    # --- plot ---
    fig, axes = plt.subplots(2, 2, figsize=(14, 12))
    

    def draw_heatmap(ax, matrix, mn9_pos, N, title):

        plot_data = np.sign(matrix) * np.log1p(np.abs(matrix))
        
        # Robust Vmax
        vals = np.abs(plot_data[plot_data!=0])
        vmax = np.percentile(vals, 99) if len(vals)>0 else 1.0
        
        im = ax.imshow(plot_data, cmap='RdBu_r', vmin=-vmax, vmax=vmax, 
                       interpolation='nearest', origin='upper')
        

        if mn9_pos is not None:
            c = 'limegreen'
            ax.axvline(mn9_pos, c=c, ls=':', alpha=0.6)
            ax.axhline(mn9_pos, c=c, ls=':', alpha=0.6)
            ax.add_patch(patches.Circle((mn9_pos, mn9_pos), N*0.03, fc='none', ec=c, lw=2.5))
            ax.text(mn9_pos+N*0.04, mn9_pos-N*0.04, "MN9", 
                    color=c, fontweight='bold', fontsize=11,
                    bbox=dict(facecolor='white', alpha=0.8, ec='none', pad=1))
            
        ax.set_title(title, fontsize=13, fontweight='bold', pad=10)
        ax.set_xticks([]); ax.set_yticks([])
        return im

    # --- Feeding Core ---
    W_f, F_f, mn9_f, N_f, _ = data_feed
    im1 = draw_heatmap(axes[0,0], W_f, mn9_f, N_f, "Feeding Core (Structure W)")
    im2 = draw_heatmap(axes[0,1], F_f, mn9_f, N_f, f"Feeding Core (Active Flux @ {feed_window[0]}-{feed_window[1]}ms)")
    
    # --- Escape Core ---
    W_e, F_e, mn9_e, N_e, _ = data_esc
    im3 = draw_heatmap(axes[1,0], W_e, mn9_e, N_e, "Escape Core (Structure W)")
    im4 = draw_heatmap(axes[1,1], F_e, mn9_e, N_e, f"Escape Core (Active Flux @ {suppress_window[0]}-{suppress_window[1]}ms)")

    # --- Colorbars  ---
    # Column 1 Colorbar (Structure)
    cbar1 = fig.colorbar(im3, ax=axes[:,0], fraction=0.05, pad=0.05, shrink=0.6, location='left')
    cbar1.set_label('Log Structural Weight', rotation=90, labelpad=15)
    
    # Column 2 Colorbar (Flux)
    cbar2 = fig.colorbar(im4, ax=axes[:,1], fraction=0.05, pad=0.05, shrink=0.6)
    cbar2.set_label('Log Signal Flux', rotation=270, labelpad=20)

    plt.suptitle("Structural Potential vs. Functional Reality", fontsize=16, y=0.95)
    
    plt.savefig(f"{save_name}.pdf", bbox_inches='tight')
    plt.savefig(f"{save_name}.png", dpi=300, bbox_inches='tight')
    print(f"✅ Saved to {save_name}.pdf/.png")
    
    plt.show()

# ==========================================
# 运行示例
# ==========================================
if __name__ == "__main__":
    try:
        compare_active_cores_final(
            sim_data, 
            feeding_core_ids=core_ids_18,   # 18个神经元
            escape_core_ids=esc_ids,        # 17个神经元
            mn9_id=MN9,
            feed_window=(1500, 2000),       # 进食态
            suppress_window=(6000, 6500),   # 抑制态
            save_name="Fig2"
        )
    except NameError:
        print("Variables not defined.")
```

```python
# Fig 4(A)
import brian2 as b2
import numpy as np
import scipy.sparse as sp
import matplotlib.pyplot as plt

def simulate_and_plot_recovery(
    neuron_ids, W_matrix, sugar_ids, threat_ids, mn9_id,
    feeding_core_ids,  
    escape_core_ids,   
    killer_ids,        
    dt_sim=0.1, 
    input_gain=200.0,
    trial_duration=6000.0, # 总时长
    ramp_down_start=1000.0, # 开始降低 Threat 的时间
    ramp_down_end=5000.0   # Threat 降为 0 的时间
):
    print(f"=== Running Single Recovery Simulation ===")
    
    b2.start_scope()
    b2.defaultclock.dt = dt_sim * b2.ms
    net = b2.Network()
    
    N = len(neuron_ids)
    id_map = {uid: i for i, uid in enumerate(neuron_ids)}
    

    try:
        idx_mn9 = id_map[mn9_id]
        idx_feed = np.array([id_map[u] for u in feeding_core_ids])
        idx_esc = np.array([id_map[u] for u in escape_core_ids])
        idx_kill = np.array([id_map[u] for u in killer_ids])
    except KeyError:
        print("Error: Some IDs not found in neuron_ids list.")
        return

    # --- 2. (Recovery) ---
    # Sugar: keep 100 Hz 
    # Threat: 200 Hz (start) -> 0 Hz (end)
    
    time_steps = np.arange(0, trial_duration + 100, dt_sim)
    sugar_vals = np.ones_like(time_steps) * 100.0 # Constant Sugar
    threat_vals = np.zeros_like(time_steps)
    
    # Threat Profile: High -> Ramp Down -> Zero
    mask_high = time_steps < ramp_down_start
    mask_ramp = (time_steps >= ramp_down_start) & (time_steps < ramp_down_end)
    
    threat_vals[mask_high] = 200.0
    threat_vals[mask_ramp] = 200.0 * (1.0 - (time_steps[mask_ramp] - ramp_down_start) / (ramp_down_end - ramp_down_start))

    
    sugar_timed = b2.TimedArray(sugar_vals*b2.Hz, dt=dt_sim*b2.ms)
    threat_timed = b2.TimedArray(threat_vals*b2.Hz, dt=dt_sim*b2.ms)
    

    tau_m = 20*b2.ms
    w_val = 0.275*b2.mV * input_gain
    
    eqs = '''
    dv/dt=(-v+g)/(20*ms):volt(unless refractory)
    dg/dt=-g/(5*ms):volt
    ref_P:second
    '''
    neurons = b2.NeuronGroup(N, eqs, threshold='v>7*mV', reset='v=0*mV', refractory='ref_P', method='exact')
    neurons.v=0; neurons.g=0; neurons.ref_P=2.2*b2.ms
    net.add(neurons)
    
    # Inputs
    s_idx = [id_map[u] for u in sugar_ids if u in id_map]
    if s_idx:
        neurons.ref_P[s_idx] = 0*b2.ms
        P_s = b2.PoissonGroup(len(s_idx), rates='sugar_timed(t)')
        S_s = b2.Synapses(P_s, neurons, on_pre='g+=w_val')
        S_s.connect(i=np.arange(len(s_idx)), j=s_idx)
        net.add(P_s, S_s)
        
    t_idx = [id_map[u] for u in threat_ids if u in id_map]
    if t_idx:
        neurons.ref_P[t_idx] = 0*b2.ms
        P_t = b2.PoissonGroup(len(t_idx), rates='threat_timed(t)')
        S_t = b2.Synapses(P_t, neurons, on_pre='g+=w_val')
        S_t.connect(i=np.arange(len(t_idx)), j=t_idx)
        net.add(P_t, S_t)
        
    # Recurrent
    if sp.issparse(W_matrix):
        coo = W_matrix.tocoo()
        S_rec = b2.Synapses(neurons, neurons, 'w:1', on_pre='g+=w*0.275*mV', delay=1.8*b2.ms)
        S_rec.connect(i=coo.col, j=coo.row)
        S_rec.w = coo.data
        net.add(S_rec)
        W_csr = W_matrix.tocsr()
    
    spikes = b2.SpikeMonitor(neurons)
    net.add(spikes)
    
    print(f"  > Simulating {trial_duration} ms...")
    net.run(trial_duration * b2.ms, namespace=locals())
    
    # --- 4. data analysis---
    print("  > Calculating Flux Dynamics...")
    all_t = spikes.t/b2.ms
    all_i = spikes.i
    
    plot_dt = 20.0
    window_ms = 100.0 # 平滑窗口
    plot_times = np.arange(0, trial_duration, plot_dt)
    

    W_cmd = W_csr[np.ix_(idx_kill, idx_esc)]
    
    # 2. Killer -> Feeding (Suppression)
    # W[idx_feed, idx_kill]
    W_supp = W_csr[np.ix_(idx_feed, idx_kill)]
    
    
    trace_cmd = []
    trace_supp = []
    trace_mn9 = []
    

    SCALE = 0.275 
    
    for t in plot_times:
        t_start = max(0, t - window_ms)
        mask = (all_t >= t_start) & (all_t < t)
        dur = window_ms / 1000.0
        
        if np.sum(mask) == 0:
            trace_cmd.append(0); trace_supp.append(0); trace_mn9.append(0)
            continue
            
        spikes_in_window = all_i[mask]
        counts = np.bincount(spikes_in_window, minlength=N)
        rates = counts / dur
        

        cmd_vec = W_cmd @ rates[idx_esc]
        trace_cmd.append(np.sum(cmd_vec) * SCALE)
        

        supp_vec = W_supp @ rates[idx_kill]
        trace_supp.append(np.sum(supp_vec) * SCALE)
        
        trace_mn9.append(rates[idx_mn9])
        
    # --- 5.plot ---
    fig, ax1 = plt.subplots(figsize=(12, 7))
    

    ax1.axvspan(ramp_down_start, ramp_down_end, color='gray', alpha=0.1, label='Threat Ramp Down')
    ax1.axhline(0, color='k', lw=0.5)
    
    # left: Flux
    l1, = ax1.plot(plot_times, trace_cmd, color='darkorange', lw=2, label='Escape Command (Esc->Kill)')
    l2, = ax1.plot(plot_times, trace_supp, color='royalblue', lw=2, label='Suppression (Kill->Feed)')
    ax1.set_ylabel('Network Flux (mV/s)', fontsize=12, fontweight='bold')
    ax1.set_xlabel('Time (ms)', fontsize=12)
    
    # right: MN9 Rate
    ax2 = ax1.twinx()
    l3, = ax2.plot(plot_times, trace_mn9, color='limegreen', lw=3, ls='--', label='MN9 Firing Rate')
    ax2.set_ylabel('MN9 Firing Rate (Hz)', color='limegreen', fontsize=12, fontweight='bold')
    ax2.tick_params(axis='y', labelcolor='limegreen')
    
    plt.title("(A) Causal sequence of recovery", fontsize=14, fontweight='bold', pad=15)
    
    # Legend
    lines = [l1, l2, l3]
    labels = [l.get_label() for l in lines]
    ax1.legend(lines, labels, loc='center right')
    
    plt.tight_layout()
    plt.savefig('Fig4(A).pdf', bbox_inches='tight')
    plt.show()

# ==========================================
# 运行示例
# ==========================================
if __name__ == "__main__":

    try:
        simulate_and_plot_recovery(
            neuron_ids=skel_ids, 
            W_matrix=skel_W, 
            sugar_ids=SUGAR, 
            threat_ids=ESC, 
            mn9_id=MN9,
            feeding_core_ids=core_ids_18, 
            escape_core_ids=esc_ids, # 来自 quantify_escape_cascade_custom 的输出
            killer_ids=kill_ids      # 来自 quantify_escape_cascade_custom 的输出
        )
    except NameError:
        print("Variables not defined. Please run previous analysis steps first.")
```

```python
# Fig 4(B)
import brian2 as b2
import numpy as np
import scipy.sparse as sp
import matplotlib.pyplot as plt


def run_single_escape_decay(
    run_id,
    neuron_ids, W_csr, threat_ids, escape_core_ids,
    dt_sim, input_gain,
    ramp_time, hold_time, decay_time, max_threat_freq,
    plot_dt=10.0, window_ms=50.0
):
    print(f"  > Trial {run_id+1} running...", end="\r")
    
    b2.start_scope()
    b2.defaultclock.dt = dt_sim * b2.ms
    
    N = len(neuron_ids)
    
    # --- Only Threat ---
    t1 = ramp_time
    t2 = t1 + hold_time # Cutoff Point
    total_time = t2 + decay_time
    
    time_steps = np.arange(0, total_time, dt_sim)
    threat_vals = np.zeros_like(time_steps)
    
    # Threat Profile
    mask_ramp = time_steps < t1
    threat_vals[mask_ramp] = max_threat_freq * (time_steps[mask_ramp] / t1)
    mask_hold = (time_steps >= t1) & (time_steps < t2)
    threat_vals[mask_hold] = max_threat_freq

    
    threat_timed = b2.TimedArray(threat_vals*b2.Hz, dt=dt_sim*b2.ms)
    

    tau_m = 20*b2.ms
    w_val = 0.275*b2.mV * input_gain
    eqs = '''
    dv/dt=(-v+g)/(20*ms):volt(unless refractory)
    dg/dt=-g/(5*ms):volt
    ref_P:second
    '''
    neurons = b2.NeuronGroup(N, eqs, threshold='v>7*mV', reset='v=0*mV', refractory='ref_P', method='exact')
    neurons.v=0; neurons.g=0; neurons.ref_P=2.2*b2.ms
    
    # Only Threat Input
    P_t = None; S_t = None
    if len(threat_ids) > 0:
        neurons.ref_P[threat_ids] = 0*b2.ms
        P_t = b2.PoissonGroup(len(threat_ids), rates='threat_timed(t)')
        S_t = b2.Synapses(P_t, neurons, on_pre='g+=w_val')
        S_t.connect(i=np.arange(len(threat_ids)), j=threat_ids)
    
    # Recurrent
    coo = W_csr.tocoo()
    S_rec = b2.Synapses(neurons, neurons, 'w:1', on_pre='g+=w*0.275*mV', delay=1.8*b2.ms)
    S_rec.connect(i=coo.col, j=coo.row); S_rec.w = coo.data
    
    spikes = b2.SpikeMonitor(neurons)
    b2.run(total_time * b2.ms, namespace=locals())
    

    all_t = spikes.t/b2.ms
    all_i = spikes.i
    plot_times = np.arange(0, total_time, plot_dt)
    
    trace_esc_rate = []
    

    
    for t in plot_times:
        t_start = max(0, t - window_ms)
        mask = (all_t >= t_start) & (all_t < t)
        dur = window_ms/1000.0
        

        if np.sum(mask) > 0:
            spikes_win = all_i[mask]
            counts_all = np.bincount(spikes_win, minlength=N)
            counts_esc = counts_all[escape_core_ids]
            avg_rate = np.mean(counts_esc) / dur
        else:
            avg_rate = 0
        trace_esc_rate.append(avg_rate)
        
    trace_esc_rate = np.array(trace_esc_rate)
    

    idx_cut = int(t2/plot_dt)
    base = np.mean(trace_esc_rate[idx_cut-20:idx_cut])
    
    if base < 1.0:
        lifetime = 0
    else:
        decay_part = trace_esc_rate[idx_cut:]
        stop_indices = np.where(decay_part < base*0.1)[0]
        if len(stop_indices) > 0:
            lifetime = plot_times[idx_cut:][stop_indices[0]] - t2
        else:
            lifetime = decay_time
            
    return plot_times, trace_esc_rate, lifetime, t2

# =============================================================================
# 2. MAIN
# =============================================================================
def measure_escape_lifetime_stats(
    neuron_ids, W_matrix, threat_ids, escape_core_ids,
    dt_sim=0.1, input_gain=200.0,
    ramp_time=500.0, hold_time=1000.0, decay_time=1000.0, max_threat_freq=200.0,
    n_trials=10
):
    print(f"=== Measuring Escape 'Quasi-Particle' Lifetime (N={n_trials}) ===")
    print(f"  > Core Size: {len(escape_core_ids)} neurons")
    

    id_map = {uid: i for i, uid in enumerate(neuron_ids)}
    try:
        t_indices = [id_map[uid] for uid in threat_ids if uid in id_map]
        esc_indices = np.array([id_map[uid] for uid in escape_core_ids])
    except KeyError:
        print("Error: IDs mismatch.")
        return

    W_csr = W_matrix.tocsr() if sp.issparse(W_matrix) else sp.csr_matrix(W_matrix)
    
    lifetimes = []
    rate_matrix = []
    common_times = None
    cutoff_time = 0
    
    for i in range(n_trials):
        t, rate, life, t_cut = run_single_escape_decay(
            i, neuron_ids, W_csr, t_indices, esc_indices,
            dt_sim, input_gain, ramp_time, hold_time, decay_time, max_threat_freq
        )
        lifetimes.append(life)
        rate_matrix.append(rate)
        if common_times is None:
            common_times = t
            cutoff_time = t_cut
            

    print(f"\n=== Escape Inertia Statistics ===")
    mean_life = np.mean(lifetimes)
    std_life = np.std(lifetimes)
    print(f"  > Lifetime: {mean_life:.0f} ± {std_life:.0f} ms")
    
    mean_rate = np.mean(np.vstack(rate_matrix), axis=0)
    std_rate = np.std(np.vstack(rate_matrix), axis=0)
    
    # --- PLOT ---
    fig, ax = plt.subplots(figsize=(10, 6))
    

    x_start = 0
    x_end = cutoff_time + mean_life + 500
    ax.set_xlim(x_start, min(x_end, common_times[-1]))
    

    ax.axvspan(0, ramp_time, color='gray', alpha=0.1, label='Threat Ramp')
    ax.axvspan(ramp_time, cutoff_time, color='red', alpha=0.05, label='High Threat State')
    ax.axvspan(cutoff_time, cutoff_time + mean_life, color='purple', alpha=0.1, label='Escape Inertia')
    ax.axvline(cutoff_time, color='k', linestyle='--', linewidth=2, label='Threat Cutoff')
    

    c_esc = 'firebrick'
    

    lower_bound = np.clip(mean_rate - std_rate, 0, None)
    
    ax.plot(common_times, mean_rate, color=c_esc, lw=3, label='Escape Core Activity')
    ax.fill_between(common_times, lower_bound, mean_rate + std_rate, 
                    color=c_esc, alpha=0.2, label='±1 SD')
    
    ax.set_xlabel('Time (ms)', fontsize=12)
    ax.set_ylabel('Escape Core Firing Rate (Hz)', color=c_esc, fontsize=12, fontweight='bold')
    ax.tick_params(axis='y', labelcolor=c_esc)
    ax.set_ylim(bottom=0)
    

    text_x = cutoff_time + mean_life / 2
    text_y = np.max(mean_rate) * 0.5
    
    ax.text(text_x, text_y, 
             f'Inertia Lifetime\n{mean_life:.0f} ± {std_life:.0f} ms', 
             ha='center', va='center', color='black', fontsize=12, fontweight='bold',
             bbox=dict(facecolor='white', alpha=0.9, edgecolor='none', pad=3))
    
    plt.title("(B) Lifetime of the escape core", fontsize=14, fontweight='bold', pad=15)
    ax.legend(loc='upper right', frameon=True)
    
    plt.tight_layout()
    plt.savefig('Fig4(B).pdf')
    plt.show()

# ==========================================
# SIMULATION
# ==========================================
if __name__ == "__main__":

    try:
        measure_escape_lifetime_stats(
            skel_ids, skel_W, ESC, esc_ids,
            n_trials=10,
            ramp_time=1000,
            hold_time=1000,
            decay_time=1000
        )
    except NameError:
        print("Variable 'esc_ids' not found. Run analyze_escape_inertia first.")
```

```python
# Fig 5
import brian2 as b2
import numpy as np
import scipy.sparse as sp
import matplotlib.pyplot as plt

def simulate_core_decay_context(
    neuron_ids, 
    W_matrix, 
    core_ids,          
    simulation_mode='isolated', # choice 'isolated' or 'embedded'
    kick_duration=20.0,    
    decay_time=2000.0,     
    kick_strength=1000.0,  
    dt_sim=0.1,
    w_scale=0.275          
):
    print(f"=== Core Simulation: {simulation_mode.upper()} Mode ===")
    
    # --- 1. data ---
    id_map = {uid: i for i, uid in enumerate(neuron_ids)}
    try:
        core_indices_global = [id_map[uid] for uid in core_ids]
    except KeyError:
        print("Error: Core IDs not found.")
        return

    b2.start_scope()
    b2.defaultclock.dt = dt_sim * b2.ms
    net = b2.Network()
    
    if simulation_mode == 'isolated':
        N = len(core_ids)
        if sp.issparse(W_matrix):
            W_sim = W_matrix[np.ix_(core_indices_global, core_indices_global)]
            coo = W_sim.tocoo()
        else: return
        target_indices = np.arange(N)
        print(f"  > Simulating {N} neurons (Isolated). Internal connections: {coo.nnz}")
        
    elif simulation_mode == 'embedded':
        N = len(neuron_ids)
        if sp.issparse(W_matrix):
            coo = W_matrix.tocoo()
        else: return
        target_indices = np.array(core_indices_global)
        print(f"  > Simulating {N} neurons (Embedded).")
    else: return

    # --- 2. equation ---
    tau_m = 20*b2.ms
    eqs = '''
    dv/dt = (-v + g) / tau_m : volt (unless refractory)
    dg/dt = -g / (5*ms) : volt
    ref_P : second
    '''
    neurons = b2.NeuronGroup(N, eqs, threshold='v>7*mV', reset='v=0*mV', refractory='ref_P', method='exact')
    neurons.v = 0 * b2.mV; neurons.g = 0 * b2.mV; neurons.ref_P = 2.2 * b2.ms
    net.add(neurons)
    
    S_rec = b2.Synapses(neurons, neurons, 'w:1', on_pre='g += w * w_scale * mV', delay=1.8*b2.ms)
    S_rec.connect(i=coo.col, j=coo.row); S_rec.w = coo.data
    net.add(S_rec)
    
    # --- 3. kick ---
    n_spikes = 10 
    num_slots = int(kick_duration / dt_sim)
    actual_spikes = min(num_slots, n_spikes)

    gen_indices = []
    gen_times = []
    
    for idx in target_indices:
        chosen_slots = np.random.choice(num_slots, actual_spikes, replace=False)
        chosen_slots.sort()
        times_ms = chosen_slots * dt_sim
        gen_indices.extend([idx] * actual_spikes)
        gen_times.extend(times_ms * b2.ms)
        
    kicker = b2.SpikeGeneratorGroup(N, gen_indices, gen_times)
    spike_strength = kick_strength / max(1, actual_spikes)
    S_kick = b2.Synapses(kicker, neurons, on_pre=f'g += {spike_strength}*mV')
    S_kick.connect(j='i')
    net.add(kicker); net.add(S_kick)


    spikes = b2.SpikeMonitor(neurons)
    net.add(spikes)
    

    total_time = kick_duration + decay_time
    print(f"  > Running ({total_time} ms)...")
    net.run(total_time * b2.ms, namespace=locals())
    
 
    all_t = spikes.t / b2.ms; all_i = spikes.i
    mask_core = np.isin(all_i, target_indices)
    core_t = all_t[mask_core]
    
    window_ms = 20.0
    bins = np.arange(0, total_time, window_ms)
    counts, _ = np.histogram(core_t, bins)
    rates = counts / (len(core_ids) * (window_ms/1000.0))
    times = bins[:-1] + window_ms/2
    
    idx_kick = int(kick_duration / window_ms)
    decay_curve = rates[idx_kick:]
    decay_times = times[idx_kick:]
    

    is_active = decay_curve > 10.0
    if np.sum(is_active) == 0:
        lifetime = 0
        status = "0.0 ms"
    else:
        if is_active[-1]:
            lifetime = decay_time 
            status = "> 2000 ms (Infinite)"
        else:
            last_idx = np.where(is_active)[0][-1]
            lifetime = decay_times[last_idx] - kick_duration
            status = f"{lifetime:.1f} ms"
            
    print(f"  > Measured Lifetime: {status}")
    
    # ---  plot---
    fig, ax = plt.subplots(figsize=(10, 5))
    
    color = 'firebrick' if simulation_mode=='isolated' else 'teal'
    ax.plot(times, rates, color=color, lw=2.5)
    ax.axvspan(0, kick_duration, color='gray', alpha=0.2, label='Kick')
    
    title_str = f"(A) Decay Dynamics: {simulation_mode.upper()} Mode"
    ax.set_title(title_str, fontsize=16, fontweight='bold', pad=15)
    ax.set_ylabel("Core Firing Rate (Hz)", fontsize=12, fontweight='bold')
    ax.set_xlabel("Time (ms)", fontsize=12)
    

    ax.text(0.5, 0.9, f"Lifetime: {status}", 
            transform=ax.transAxes, 
            ha='center', va='top', 
            fontsize=14, fontweight='bold', 
            bbox=dict(facecolor='white', alpha=0.9, edgecolor='black', boxstyle='round,pad=0.4'))
    
    ax.legend(loc='upper right', frameon=True)
    plt.tight_layout()
    plt.savefig(f'Figure_Decay_{simulation_mode}.pdf')
    plt.show()

# ==========================================
# run
# ==========================================
if __name__ == "__main__":
    try:
        simulate_core_decay_context(skel_ids, skel_W, core_ids_18, simulation_mode='isolated')
        simulate_core_decay_context(skel_ids, skel_W, core_ids_18, simulation_mode='embedded')
    except NameError:
        pass
```
